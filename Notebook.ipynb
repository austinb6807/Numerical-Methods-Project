# Euler's Method, Improved Euler's Method, and Runge Kutta 4 Method

With the governing equation in differential form, the problem takes the form of an Initial Value Problem, where the state of the system can be defined given its initial values to create a unique solution. Above, we saw the closed-form solution to this IVP. Because the solution can be explicitly written, we can compare the exact solution to numerical solutions found with Euler's Method, the Improved Euler's Method, and the Runge Kutta 4 Method.

For an initial value problem, we define the derivative of the function as $ y' = f \left( t, y \right) $. At any given point in the system, $\left(t, y \right) $ is the state of the system, where t is an independent variable and y is a dependent variable. To close the initial value problem, we denote $y_0$ and $t_0$ as the initial values of the problem. For a $n$ order system, the system is defined with $n$ initial state points defined.

For numerical methods applied to IVPs, we iterate through the state of the system by "stepping" from the initial state to a second state, and so on. We define the timestep, $h$, such that $ t_{n+1} = t_n + h $, which allows us to approximate the exact solution $ y \left( t_n \right) $. By discretizing the system, we can approdimate the solution with methods such that $ y \left( t_n \right) \approx f \left(t_{n -1}, y_{n-1} \right) h + y_{n-1}$.

**Big O notation** describes how a method's error scales as the step size decreases. We write $E(h)=O(h^p)$, which means that the error is on the order of $h^p$, or $|E(h)|\leq C h^p$ for sufficiently small $h$, where $C$ is a constant. A higher exponent value $p$ indicates that error decreases more rapidly as $h$ is decreased.

Below is a table describing the big O behavior for the three numerical methods that we will explore. These methods and their error order will be explained subsequently.


| Method | Error Order | Slope evaluations per step |
| --- | --- | --- |
| Euler | $O(h)$ | 1 |
| Improved Euler | $O(h^2)$ | 2 |
|  Runge Kutta 4 | $O(h^4)$ | 4 |

In this project, we will compare the error for each method to the exact solution of our convection-conduction problem through the pipe. We will do so to verify that the numerical methods are working as expected, i.e. the error orders are approximately their expected theoretical values, as well as to compare their computational cost.

# Euler's Method #

Euler's Method uses the slope at the beginning of each step to predict the next state:

$$
y_{n+1}=y_n + f(t_n, y_n)h
$$

In essence, the method follows the tangent to the solution curve over a step of size $h$. It requires one evaluation of the differential equation per step.

To determine its accuracy, we utilize taylor polynomial expansion to expand the exact solution about $t_n$, and observe how many terms of the taylor polynomial align with this numerical method. The taylor polynomial expansion is:

$$
y (t_n+h)=y (t_n)+h y'(t_n)+\frac{h^2}{2}y''(t_n)+ ... 
$$

Since $y'=f(t,y)$, Euler's method retains the zero-th order temr and the first order term. Starting from the exact state, the first omitted term is $\frac{h^2}{2}\mathbf{y}''(t_n)$, so its accuracy is on the order of $h$, denoted in big O notation as $ O \left( h \right). This makes Euler's method a **first-order method**.

Because Euler's Method only utilizes one slope evaluation per step, it is per step inexpensive, but requires smaller step sizes than higher-order methods to achieve a given accuracy.

# Improved Euler's Method #

Improved Euler's Method uses the slope at the beginning of each step and the slope at a predicted endpoint to approximate the next state. First, we utilize Euler's Method to predict the endpoint, then average the two slopes to calculate the next state:

$$
\begin{aligned}
k_1 &= f(t_n,y_n),\\
\widetilde{y}_{n+1} &= y_n+h k_1,\\
k_2 &= f(t_n+h,\widetilde{y}_{n+1}),\\
y_{n+1} &= y_n+\frac{h}{2}(k_1+k_2).
\end{aligned}
$$

In essence, the method accounts for how the slope changes across a step of size $h$, rather than only using the initial slope. It requires two evaluations of the differential equation per step. This version of Improved Euler's Method is also known as Heun's Method.

To determine its accuracy, we utilize Taylor polynomial expansion to expand the exact solution about $t_n$, and observe how many terms of the Taylor polynomial align with this numerical method. The Taylor polynomial expansion is:

$$
y(t_n+h)=y(t_n)+h y'(t_n)+\frac{h^2}{2}y''(t_n)+\frac{h^3}{6}y'''(t_n)+\cdots
$$

Since $y'=f(t,y)$, we can expand the predicted endpoint slope about the initial state:

$$
k_2=f(t_n,y_n)+h\left(\frac{\partial f}{\partial t}+f\frac{\partial f}{\partial y}\right)+\cdots
$$

The derivatives on the right are evaluated at $(t_n,y_n)$. By the chain rule, $y''=\frac{\partial f}{\partial t}+f\frac{\partial f}{\partial y}$. Substituting the slope expansion into the update gives, when starting from the exact state,

$$
y_{n+1}=y(t_n)+h y'(t_n)+\frac{h^2}{2}y''(t_n)+\text{terms of order }h^3\text{ and higher}.
$$

Improved Euler's Method therefore retains the zero-th order term, the first order term, and the second order term of the exact Taylor expansion. Over a fixed integration interval, its error is on the order of $h^2$, denoted in big O notation as $O(h^2)$. This makes Improved Euler's Method a **second-order method**.

Because Improved Euler's Method utilizes two slope evaluations per step, it is more expensive per step than Euler's Method, but generally allows for larger step sizes to achieve a given accuracy.

# Runge Kutta 4

The Runge Kutta 4 Method uses four slope evaluations to predict the next state: one at the beginning of the step, two at predicted states halfway through the step, and one at a predicted endpoint:

$$
\begin{aligned}
k_1 &= f(t_n,y_n),\\
k_2 &= f\left(t_n+\frac{h}{2},y_n+\frac{h}{2}k_1\right),\\
k_3 &= f\left(t_n+\frac{h}{2},y_n+\frac{h}{2}k_2\right),\\
k_4 &= f(t_n+h,y_n+h k_3).
\end{aligned}
$$

We then utilize a weighted average of these slopes to calculate the next state:

$$
y_{n+1}=y_n+\frac{h}{6}(k_1+2k_2+2k_3+k_4).
$$

In essence, the method uses information from the beginning, middle, and end of a step of size $h$ to account for changes in the slope. It requires four evaluations of the differential equation per step, without requiring us to calculate higher derivatives directly.

To determine its accuracy, we utilize Taylor polynomial expansion to expand the exact solution about $t_n$, and observe how many terms of the Taylor polynomial align with this numerical method. The Taylor polynomial expansion is:

$$
y(t_n+h)=y(t_n)+h y'(t_n)+\frac{h^2}{2}y''(t_n)+\frac{h^3}{6}y'''(t_n)+\frac{h^4}{24}y^{(4)}(t_n)+\cdots
$$

When starting from the exact state, expanding the four slopes about the initial state and combining them with the weights above gives:

$$
\frac{k_1+2k_2+2k_3+k_4}{6}=y'(t_n)+\frac{h}{2}y''(t_n)+\frac{h^2}{6}y'''(t_n)+\frac{h^3}{24}y^{(4)}(t_n)+\text{terms of order }h^4\text{ and higher}.
$$

Multiplying this expression by $h$ and adding the initial state reproduces the zero-th order term through the fourth order term of the exact Taylor expansion. Over a fixed integration interval, its error is on the order of $h^4$, denoted in big O notation as $O(h^4)$. This makes the Runge Kutta 4 Method a **fourth-order method**.

Because the Runge Kutta 4 Method utilizes four slope evaluations per step, it is more expensive per step than Euler's Method and Improved Euler's Method, but generally allows for larger step sizes to achieve a given accuracy. The step size must still be small enough for numerical stability.
